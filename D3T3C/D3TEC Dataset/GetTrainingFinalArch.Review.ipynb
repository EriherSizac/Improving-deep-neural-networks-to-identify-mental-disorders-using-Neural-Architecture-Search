```python
import pickle
import os

checkpoint_path = 'F:/Github/Improving-deep-neural-networks-to-identify-mental-disorders-using-Neural-Architecture-Search/D3T3C/D3TEC Dataset/checkpoints/saves_20250424_125632/final_checkpoint.pkl'
with open(checkpoint_path, 'rb') as file:
    final_checkpoint = pickle.load(file)
print(final_checkpoint)

all_F = []
all_fitness = []
all_best_architectures = []

for i in range(30):  # Adjust the range based on the number of experiments conducted
    checkpoint_file = f"F:/Github/Improving-deep-neural-networks-to-identify-mental-disorders-using-Neural-Architecture-Search/D3T3C/D3TEC Dataset/checkpoints/saves_20250423_231616/final_checkpoint_exp{i}.pkl"
    if os.path.exists(checkpoint_file):
        with open(checkpoint_file, 'rb') as file:
            checkpoint = pickle.load(file)
            print(checkpoint)
            print(checkpoint['F_history'])
            print(checkpoint['fitness_history'])
            print(checkpoint['best_model_exp']['individual'])
            break
        all_F.extend(checkpoint['F_history'])
        all_fitness.extend(checkpoint['fitness_history'])
        all_best_architectures.append(checkpoint['best_model_exp']['individual'])

# Print the results
print(f"All F: {all_F}")
print(f"All Fitness: {all_fitness}")
print(f"All Best Architectures: {all_best_architectures}")
```
