```python
import pickle
import os

checkpoint_path = f'F:/Github/Improving-deep-neural-networks-to-identify-mental-disorders-using-Neural-Architecture-Search/D3T3C/D3TEC Dataset/checkpoints/{SAVE}/final_checkpoint.pkl'
with open(checkpoint_path, 'rb') as file:
    final_checkpoint = pickle.load(file)
print(final_checkpoint)

all_F = []
all_fitness = []
all_best_architectures = []

for i in range(30):  # Adjust the range based on the number of experiments conducted
    checkpoint_file = f"F:/Github/Improving-deep-neural-networks-to-identify-mental-disorders-using-Neural-Architecture-Search/D3T3C/D3TEC Dataset/checkpoints/{SAVE}/final_checkpoint_exp{i}.pkl"
    if os.path.exists(checkpoint_file):
        with open(checkpoint_file, 'rb') as file:
            checkpoint = pickle.load(file)
            """    print(checkpoint)
            print(checkpoint['F_history'])
            print(checkpoint['fitness_history'])
            print(checkpoint['best_model_exp']['individual']) """
            all_F.extend(checkpoint['F_history'])
            all_fitness.extend(checkpoint['fitness_history'])
            all_best_architectures.append(checkpoint['best_model_exp']['individual'])

# Print the results
print(f"All F: {all_F}")
print(f"All Fitness: {all_fitness}")
print(f"All Best Architectures: {all_best_architectures}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Reorganiza los datos para el gráfico de fitness (F1)
# Primero, necesitamos reorganizar all_F en una lista de listas (por experimento)
all_f1_by_exp = []
F_arr = []  # Para el segundo gráfico

for i in range(30):  # Asumiendo 30 experimentos
    checkpoint_file = f"F:/Github/Improving-deep-neural-networks-to-identify-mental-disorders-using-Neural-Architecture-Search/D3T3C/D3TEC Dataset/checkpoints/{SAVE}/final_checkpoint_exp{i}.pkl"
    if os.path.exists(checkpoint_file):
        with open(checkpoint_file, 'rb') as file:
            checkpoint = pickle.load(file)
            # Guarda el historial de F1 para este experimento
            all_f1_by_exp.append(checkpoint['fitness_history'])
            # Guarda el historial de F para este experimento
            F_arr.append(checkpoint['F_history'])

# Parámetros
n = 10       # Intervalo de adaptación (cada n generaciones)
num_runs = len(all_f1_by_exp)

# --- Convergencia de F1 por generación ---
# Máximo número de generaciones
max_gens = max(len(run) for run in all_f1_by_exp)
gens = np.arange(1, max_gens + 1)

# Construye la matriz y rellena con NaN
f1_mat = np.full((num_runs, max_gens), np.nan)
for i, run in enumerate(all_f1_by_exp):
    f1_mat[i, :len(run)] = run

plt.figure(figsize=(12, 6))
for i in range(num_runs):
    plt.plot(gens, f1_mat[i], color='gray', alpha=0.5)

mean_f1 = np.nanmean(f1_mat, axis=0)
std_f1 = np.nanstd(f1_mat, axis=0)

plt.plot(gens, mean_f1, color='blue', linewidth=2, label='F1 mean')
plt.fill_between(gens, mean_f1 - std_f1, mean_f1 + std_f1,
                 color='blue', alpha=0.2, label='Std dev')

plt.title(f'Convergence of F1 per generation ({num_runs} runs)')
plt.xlabel('Generation')
plt.ylabel('F1 Score')
plt.legend()
plt.grid(True)
plt.show()

# --- Evolución de F cada n generaciones ---
plt.figure(figsize=(12, 6))

for Fs in F_arr:
    # Generaciones donde se mide F
    gens_F = [n * i for i in range(1, len(Fs) + 1)]
    # Asegura que el último punto sea la última generación
    if gens_F and gens_F[-1] != max_gens:
        gens_F[-1] = max_gens
    plt.plot(gens_F, Fs, color='red', alpha=0.3)

# Media y padding (solo si hay datos)
if F_arr:
    max_len = max(len(Fs) for Fs in F_arr)
    # Padding con el último valor
    Fs_pad = [Fs + [Fs[-1]] * (max_len - len(Fs)) for Fs in F_arr if Fs]
    
    if Fs_pad:
        mean_F = np.mean(Fs_pad, axis=0)
        mean_gens = [n * i for i in range(1, max_len + 1)]
        if mean_gens and mean_gens[-1] != max_gens:
            mean_gens[-1] = max_gens
        
        plt.plot(mean_gens, mean_F, color='black', linewidth=2, label='F mean')

plt.title(f'Evolution of F every {n} generations')
plt.xlabel('Generation')
plt.ylabel('F value')
plt.legend(loc='upper right')
plt.grid(True)
plt.show()
```
